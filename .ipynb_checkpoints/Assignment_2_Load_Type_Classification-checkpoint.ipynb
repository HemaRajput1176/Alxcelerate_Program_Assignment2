# Load Type Prediction — USEReady Assignment

**Objective:** Predict `Load_Type` (`Light_Load`, `Medium_Load`, `Maximum_Load`) from historical power-system readings.

This notebook uses chronological validation: November is used for model selection and the final calendar month is held out for testing. Each analysis step is kept in its own cell, followed by a short plain-English interpretation.

```python
from pathlib import Path
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler
from sklearn.dummy import DummyClassifier
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier, HistGradientBoostingClassifier
from sklearn.metrics import (
    accuracy_score, precision_recall_fscore_support, f1_score,
    classification_report, ConfusionMatrixDisplay, confusion_matrix
)
```

## 1. Load the data

Keep the CSV in the Colab session, or update the path below to match its location.

```python

RANDOM_STATE = 42
DATA_PATH = Path('/content/load_data.csv')
if not DATA_PATH.exists():
    DATA_PATH = Path('load_data.csv')
if not DATA_PATH.exists():
    from google.colab import files
    uploaded = files.upload()
    DATA_PATH = Path(next(iter(uploaded)))

df = pd.read_csv(DATA_PATH)
print(f"Rows: {len(df):,} | Columns: {df.shape[1]}")
print("Columns:", df.columns.tolist())
display(df.head())
```

**What this tells us**

The first rows confirm that the file loaded and show the available fields. Next, we check data types and quality before changing any values.

## 2. Check data quality

Review column types, missing values, and exact duplicate rows.

```python
df.info()
```

```python

missing = pd.DataFrame({
    'missing_count': df.isna().sum(),
    'missing_percent': (df.isna().mean() * 100).round(2)
})
display(missing.sort_values('missing_percent', ascending=False))
print('Exact duplicate rows:', df.duplicated().sum())
```

**What this tells us**

Missing values should be handled without using information from the test month. We will keep rows where possible and let training-fitted imputers handle missing feature values.

## 3. Parse timestamps and inspect repeated times

The timestamp is needed to create a chronological train/validation/test split.

```python
df['Date_Time'] = pd.to_datetime(
    df['Date_Time'], format='%d-%m-%Y %H:%M', errors='coerce'
)
print('Unparsed timestamps:', df['Date_Time'].isna().sum())
df = df.dropna(subset=['Date_Time']).copy()

dup_mask = df['Date_Time'].duplicated(keep=False)
print('Rows sharing a timestamp:', int(dup_mask.sum()))
display(df.loc[dup_mask].sort_values('Date_Time'))
print('Date range:', df['Date_Time'].min(), 'to', df['Date_Time'].max())
```

**What this tells us**

Repeated timestamps need inspection because the same time should not accidentally count twice. If one duplicate is clearly more complete and physically plausible, so we need to keep that record; do not remove unrelated rows just because they have missing values.

Observed: There is exactly one duplicate timestamp (2018-12-31 20:00:00). Row 35024 contains an invalid Lagging_Current_Power_Factor value of 217.27 and missing (NaN) Usage_kWh, whereas row 35025 has valid data (Usage_kWh = 4.14, power factor = 100.00).

## 4. Resolve the duplicate timestamp and inspect invalid power factors

Power-factor values above 100 are outside the usual percentage scale. We mark those measurements as missing rather than treating them as valid readings.

```python
target = 'Load_Type'
lag_pf = 'Lagging_Current_Power_Factor'
lead_pf = 'Leading_Current_Power_Factor'

# Prefer rows with a measured target feature and plausible power factors.
df['_quality_score'] = df['Usage_kWh'].notna().astype(int)
for col in [lag_pf, lead_pf]:
    df['_quality_score'] += (df[col].between(0, 100) | df[col].isna()).astype(int)
df = (df.sort_values(['Date_Time', '_quality_score'], ascending=[True, False])
        .drop_duplicates(subset='Date_Time', keep='first')
        .drop(columns='_quality_score')
        .reset_index(drop=True))

print('Rows after timestamp deduplication:', len(df))
for col in [lag_pf, lead_pf]:
    invalid = ~df[col].between(0, 100) & df[col].notna()
    print(f'{col}: {int(invalid.sum())} values outside 0–100')
    df.loc[invalid, col] = np.nan

print('Missing Usage_kWh retained for pipeline imputation:', int(df['Usage_kWh'].isna().sum()))
```

**What this tells us**

The duplicate is resolved by keeping the more plausible record, and out-of-range power factors are treated as missing. The imputer will learn replacement values from training data only.

## 12. Prepare time-based features

Use timestamp-derived features and exclude raw `NSM` because it disagrees with the timestamp in some records. The timestamp itself is not passed directly to the model.

```python
df = df.sort_values('Date_Time').reset_index(drop=True)
minutes = df['Date_Time'].dt.hour * 60 + df['Date_Time'].dt.minute

df['time_sin'] = np.sin(2 * np.pi * minutes / 1440)
df['time_cos'] = np.cos(2 * np.pi * minutes / 1440)
df['day_of_week'] = df['Date_Time'].dt.dayofweek
df['month'] = df['Date_Time'].dt.month

# NSM should match seconds since midnight; report mismatches for transparency.
expected_nsm = df['Date_Time'].dt.hour * 3600 + df['Date_Time'].dt.minute * 60
nsm_mismatch = (df['NSM'] - expected_nsm).abs() > 60
print('NSM values inconsistent with timestamp:', int(nsm_mismatch.sum()))
print('Time features created: time_sin, time_cos, day_of_week, month')
```

**What this tells us**

The cyclical features represent times near midnight as close together instead of far apart. We use the timestamp as the source for time features and leave the inconsistent raw `NSM` out of the model.

## 13. Create chronological splits

The latest calendar month is the test set. November is used for model selection, and earlier records are used for training.

```python
last_month = df['Date_Time'].max().to_period('M')
test_start = last_month.start_time
val_start = test_start - pd.offsets.MonthBegin(1)

train_df = df[df['Date_Time'] < val_start].copy()
val_df = df[(df['Date_Time'] >= val_start) & (df['Date_Time'] < test_start)].copy()
test_df = df[df['Date_Time'] >= test_start].copy()

for name, part in [('Train', train_df), ('Validation', val_df), ('Test', test_df)]:
    print(f'{name}: {len(part):,} rows | {part.Date_Time.min()} to {part.Date_Time.max()}')

assert train_df['Date_Time'].max() < val_df['Date_Time'].min()
assert val_df['Date_Time'].max() < test_df['Date_Time'].min()
assert test_df['Date_Time'].dt.to_period('M').nunique() == 1
```

**What this tells us**

The split keeps future observations out of training and reserves the latest month for the final check. We will use validation results to choose the model; the test set will not guide that choice.

## 14. Define features and preprocessing

Keep only predictors that are available at prediction time. Imputation and scaling are fitted inside each model pipeline.

```python
feature_cols = [
    'Usage_kWh',
    'Lagging_Current_Reactive.Power_kVarh',
    'Leading_Current_Reactive_Power_kVarh',
    'CO2(tCO2)', lag_pf, lead_pf,
    'time_sin', 'time_cos', 'day_of_week', 'month'
]

X_train, y_train = train_df[feature_cols], train_df[target]
X_val, y_val = val_df[feature_cols], val_df[target]
X_test, y_test = test_df[feature_cols], test_df[target]

numeric_transformer = Pipeline([
    ('imputer', SimpleImputer(strategy='median'))
])
scaled_transformer = Pipeline([
    ('imputer', SimpleImputer(strategy='median')),
    ('scaler', StandardScaler())
])

print('Features:', feature_cols)
print('Missing target labels - train/validation/test:',
      y_train.isna().sum(), y_val.isna().sum(), y_test.isna().sum())
```

**What this tells us**

The target is excluded from the inputs, and missing feature values are handled within each pipeline. Before interpreting scores, confirm that every chosen feature would genuinely be available when a prediction is made.

## 15. Build candidate models

Compare a simple baseline with a linear model and two tree-based models. All candidates use the same chronological validation period.

```python
logistic_preprocessor = ColumnTransformer([
    ('scaled_numeric', scaled_transformer, feature_cols)
])
tree_preprocessor = ColumnTransformer([
    ('numeric', numeric_transformer, feature_cols)
])

models = {
    'Majority baseline': DummyClassifier(strategy='most_frequent'),
    'Logistic Regression': Pipeline([
        ('prep', logistic_preprocessor),
        ('model', LogisticRegression(max_iter=2000, random_state=RANDOM_STATE))
    ]),
    'Random Forest': Pipeline([
        ('prep', tree_preprocessor),
        ('model', RandomForestClassifier(
            n_estimators=250, min_samples_leaf=2,
            random_state=RANDOM_STATE, n_jobs=-1
        ))
    ]),
    'Hist Gradient Boosting': Pipeline([
        ('prep', tree_preprocessor),
        ('model', HistGradientBoostingClassifier(
            max_iter=150, learning_rate=0.08, random_state=RANDOM_STATE
        ))
    ])
}
```

we define and prepare four candidate models (Majority baseline, Logistic Regression, Random Forest, and Hist Gradient Boosting) within sklearn pipelines. This step sets up automated imputation and scaling for each model type, ensuring that data transformations are cleanly isolated and fitted only on the training periods to avoid feature leakage.

## 16. Evaluate candidates on validation data

```python
validation_results = []
for name, model in models.items():
    model.fit(X_train, y_train)
    pred = model.predict(X_val)
    validation_results.append({
        'Model': name,
        'Accuracy': accuracy_score(y_val, pred),
        'Macro-F1': f1_score(y_val, pred, average='macro'),
        'Weighted-F1': f1_score(y_val, pred, average='weighted')
    })

validation_results = pd.DataFrame(validation_results).sort_values('Macro-F1', ascending=False)
display(validation_results.round(4))
```

**What this tells us**

Use the validation table to compare models on the same unseen month. Macro-F1 gives each class equal weight, so choose the final model using the metric that best matches the assignment rather than accuracy alone.

## 17. Visualize validation performance

```python
plot_data = validation_results.melt(
    id_vars='Model', value_vars=['Accuracy', 'Macro-F1'],
    var_name='Metric', value_name='Score'
)
plt.figure(figsize=(9, 4))
sns.barplot(data=plot_data, x='Model', y='Score', hue='Metric')
plt.ylim(0, 1)
plt.title('Model Comparison on November Validation Data')
plt.xticks(rotation=15, ha='right')
plt.tight_layout()
plt.show()
```

**What this tells us**

This chart makes the accuracy and macro-F1 trade-off easier to compare. We will select the model with the strongest validation macro-F1, then refit it using all data available before the test month.

## 18. Select a model using validation only

No test-set score is used to select the model.

```python
best_name = validation_results.iloc[0]['Model']
print('Selected model:', best_name)
print('Selection metric: validation Macro-F1')
print('Validation Macro-F1:', round(validation_results.iloc[0]['Macro-F1'], 4))
```

**What this tells us**

The selected model is the best candidate by validation macro-F1 among those tested. This is a comparison of these candidates, not proof that no other model could perform better.

## 19. Fit the selected model on all pre-test data

After model selection, combine training and validation data. The final month remains untouched until the next step.

```python
pretest_df = pd.concat([train_df, val_df]).sort_values('Date_Time')
X_pretest, y_pretest = pretest_df[feature_cols], pretest_df[target]
final_model = models[best_name]
final_model.fit(X_pretest, y_pretest)
print(f'Final model fitted on {len(pretest_df):,} rows before {test_start.date()}.')
```

**What this tells us**

The final model now uses all historical rows before the test month. Its settings were chosen earlier, so the December results can serve as a final evaluation.

## 20. Final test evaluation — latest month only

```python
test_preds = final_model.predict(X_test)
print('Test month:', str(last_month))
print('Accuracy:', round(accuracy_score(y_test, test_preds), 4))
print('Macro-F1:', round(f1_score(y_test, test_preds, average='macro'), 4))
print('Weighted-F1:', round(f1_score(y_test, test_preds, average='weighted'), 4))
print('\nClassification report:')
print(classification_report(y_test, test_preds, digits=3, zero_division=0))
```

**What this tells us**

These are the final scores for the latest month, which was not used for model selection. Compare the class-level scores as well as the overall metrics to see which load types are harder to predict.

## 21. Confusion matrix

```python
labels = ['Light_Load', 'Medium_Load', 'Maximum_Load']
cm = confusion_matrix(y_test, test_preds, labels=labels)
display_cm = ConfusionMatrixDisplay(confusion_matrix=cm, display_labels=labels)
fig, ax = plt.subplots(figsize=(7, 6))
display_cm.plot(ax=ax, cmap='Blues', values_format='d', colorbar=False)
plt.title(f'Confusion Matrix — {last_month} Test Set')
plt.xticks(rotation=20, ha='right')
plt.tight_layout()
plt.show()
```

**What this tells us**

The diagonal cells are correct predictions; off-diagonal cells show which classes the model confuses. Focus on the largest off-diagonal counts when describing the model's remaining errors.

## Review misclassified examples

```python
error_review = test_df[['Date_Time', target] + feature_cols].copy()
error_review['Predicted_Load_Type'] = test_preds
misclassified = error_review[error_review[target] != error_review['Predicted_Load_Type']]
print(f'Misclassified: {len(misclassified):,} of {len(test_df):,} rows '
      f'({len(misclassified) / len(test_df):.1%})')
display(misclassified[['Date_Time', 'Usage_kWh', target, 'Predicted_Load_Type']].head(10))
```

```python
train_preds = final_model.predict(X_pretest)

print('--- Training Set (Pre-test) Performance ---')
print('Accuracy:', round(accuracy_score(y_pretest, train_preds), 4))
print('Macro-F1:', round(f1_score(y_pretest, train_preds, average='macro'), 4))

print('\n--- Test Set (December) Performance ---')
print('Accuracy:', round(accuracy_score(y_test, test_preds), 4))
print('Macro-F1:', round(f1_score(y_test, test_preds, average='macro'), 4))

# Calculate the generalization gap
acc_gap = accuracy_score(y_pretest, train_preds) - accuracy_score(y_test, test_preds)
f1_gap = f1_score(y_pretest, train_preds, average='macro') - f1_score(y_test, test_preds, average='macro')
print(f'\nGeneralization Gap (Train - Test):')
print(f'Accuracy Gap: {acc_gap:.4f}')
print(f'Macro-F1 Gap: {f1_gap:.4f}')
```

**checking that model is overfitting or not**

These examples show where predictions differ from the recorded labels. Treat explanations for why an error occurred as hypotheses unless the data directly supports them.

> Based on the execution results, the model shows a small amount of expected overfit, but remains highly robust:

> Training Set: Accuracy of 99.70% (Macro-F1 of 99.71%)
Test Set (December): Accuracy of 93.99% (Macro-F1 of 93.31%)

> **Conclusion**: While there is a minor generalization gap of ~5.7% in accuracy and ~6.4% in Macro-F1, this is perfectly normal for a complex tree-based ensemble (Random Forest) operating on seasonal time-series data. A test accuracy of nearly 94% on completely unseen future data indicates that the model has successfully generalized and is not severely overfitted.
